# Group 9 DQN: run experiments on Kaggle

Runs your experiments from `experiments/<Name>/run_experiments.sh` on a Kaggle GPU. It does the same as `bash kaggle/launch.sh`, but from the Kaggle website, so you don't need the command line.

**Before running**
1. In the right-hand panel (**Settings**, or **Session options**): set **Accelerator** to a **GPU T4** option and turn **Internet** on. Internet needs a verified phone number on your Kaggle account.
2. Edit the three settings in the next cell.
3. Click **Save Version**, choose **Save & Run All (Commit)**, then **Save**. It runs in the background for up to 12 hours, so you can close the tab.
4. When the version has finished, open it, go to **Output**, and download `<Name>_results.zip`. Extract it in the repo folder so the files land in `experiments/<Name>/exp01`, `exp02`, and so on.

exp01–exp09 take about 4.5 hours in total. See `KAGGLE_SETUP.md` in the repo for how to commit the results.

```python
MEMBER = "Homere"            # Hassan, Homere or Kevin: exactly as the folder name in experiments/
EXPERIMENTS = "1-9"         # "1-9" (exp01-exp09), "10", "2,5", or "check" (setup and GPU test only)
GIT_REF = "v1-experiments"  # "v1-experiments" for exp01-exp09; "main" (or your branch) for exp10
```

```python
# Download the group's Kaggle runner (kaggle/job.py) at GIT_REF, fill in the settings, and run it.
# It installs uv, clones the repo, installs the locked packages, checks the GPU, then runs the selected
# lines of experiments/<MEMBER>/run_experiments.sh one after another, copying each finished
# experiment folder to /kaggle/working (the notebook's Output).
import urllib.request

url = f"https://raw.githubusercontent.com/Hassan-Adelani-Luqman/group-9-atari-dqn/{GIT_REF}/kaggle/job.py"
job = urllib.request.urlopen(url).read().decode()
for key, value in {"MEMBER": MEMBER, "EXPERIMENTS": EXPERIMENTS, "GIT_REF": GIT_REF}.items():
    job = job.replace("{{" + key + "}}", value)
with open("/tmp/job.py", "w") as file:
    file.write(job)

!python /tmp/job.py
```

```python
# Put all results in one zip file, so they can be downloaded from the Output tab in one click.
import shutil
from pathlib import Path

if Path("/kaggle/working/experiments").exists():
    archive = shutil.make_archive(f"/kaggle/working/{MEMBER}_results", "zip", "/kaggle/working", "experiments")
    print("Done. Download this file from the Output tab:", archive)
    for folder in sorted(Path("/kaggle/working/experiments").glob("*/exp*")):
        print(" ", folder.relative_to("/kaggle/working"), "->", sorted(p.name for p in folder.iterdir()))
else:
    print("No experiment results (setup check, or the run failed: read the output of the cell above).")
```
